```python
import pickle
import numpy as np
```

```python
oe = pickle.load(open("models/oe.pkl","rb"))
ohe = pickle.load(open("models/ohe.pkl","rb"))
clf = pickle.load(open("models/clf.pkl","rb"))
```

```python
# assume our input is:
# Sex	Age	ChestPainType	Severity	ST_Slope	RestingECG	
#  M    58	    ASY          Low	       Up	     Normal
```

```python
test_input = np.array(['M',58,'ASY','Low','Up','Normal'],dtype=object).reshape(1,6)
```

```python
test_input[:,[0,2,5]]
test_input[:,[4,3]]
```

# Note: <br>
### It is very critical to know the sequence by which we have passed our columns in our encoders and in X_train as well when fitting the model. The same sequence must be followed when passing the columns during production.

```python
transformed_ordinal_cols = oe.transform(test_input[:,[4,3]])
transformed_nominal_cols = ohe.transform(test_input[:,[0,2,5]])
transformed_test_input = np.concatenate((test_input[:,1].reshape(-1,1),transformed_ordinal_cols,transformed_nominal_cols),axis=1)
transformed_test_input
```

```python
clf.predict(transformed_test_input)
```
