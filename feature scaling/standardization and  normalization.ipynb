## Building Feature Scaling Formulas from scratch

    Write two functions,
    standardize(data) and normalize(data),
    that take a list/array of numbers and return the scaled version.
    use only basic arithmetic (no sklearn, no numpy .std() shortcuts.
if you can help it, compute mean/std manually first).<br>
Test both on [4, 8, 15, 16, 23, 42].
</br>
Requirement:
    <li>your standardize function should give a result with mean ≈ 0 and std ≈ 1.</li>
    <li>verify that numerically after the fact, don't just trust the formula.</li>

```python
class mylist:
    @classmethod
    def sum(cls,data):
        total = 0 
        for itr in data:     # data = [4,8,15,16,23,42]
            total = itr + total
        return total
    @classmethod
    def mean(cls,data):
            return cls.sum(data) / len(data)
    @classmethod
    def std_dev(cls,data): 
        m = cls.mean(data)
        sum_of_sq = 0 
        for itr in data:
            sum_of_sq = (itr - m)** 2 + sum_of_sq
        return  (((sum_of_sq)/ len(data))** (1/2))
    @classmethod
    def max(cls,data): # i could have used python's built in min-max methods for lists. but since all its aimed at learnin, 
                       # i have played carefully not to overshadow or conflict with exisitng methods and by defining things similarily,
                       #  i can imitate how that might be implement in actual libraries.
        max = data[0]
        for itr in data:
            if (itr >= max):
                max = itr 
        return max
    @classmethod
    def min(cls,data):
        min = data[0]
        for itr in data:
            if (itr <= min):
                min = itr
        return min 
            
        
```

```python
class feature_scaling:
    @staticmethod
    def standardize(list): # this uses z-score to scale values.
        scaled_list = []
        mean = mylist.mean(list)
        std_dev = mylist.std_dev(list)
        for itr in list:
            scaled_list.append((itr - mean )/ std_dev)
        return scaled_list
    @staticmethod       
    def normalize(data): # it uses min-max normalization technique
        normalized_list = []
        min = mylist.min(data)
        max = mylist.max(data)
        diff = max - min 
        for itr in data:
            normalized_list.append((itr - min)/diff)
        return normalized_list

        
```

```python
scaler = feature_scaling()
```

```python
scaled_list = scaler.standardize( [4,8,15,16,23,42])
print(scaled_list)
```

```python
normalized_list = scaler.normalize( [4,8,15,16,23,42])
print(normalized_list)
```

```python
print('std dev =',mylist.std_dev(scaled_list)) # as learned, the standard deviation of scaled list becomes 1 
print('mean = ',mylist.mean(scaled_list))   # and mean becomes 0.
```

```python
print('std dev =',mylist.std_dev(normalized_list)) # as learned, all the values are scaled between 0 and 1.
print('mean = ',mylist.mean(normalized_list))   # but mean is not necessarily centered or mean = 0
```
