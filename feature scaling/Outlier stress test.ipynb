Take a small pandas DataFrame with 2-3 numeric columns of very different scales<br> (e.g., age, income, purchased make up plausible values, ~10 rows).<br>Apply sklearn.preprocessing.StandardScaler and MinMaxScaler separately. <br>Requirement:<br> before looking at any output, predict what .mean() and .std() of each column should be <br>after StandardScaler, and what .min()/.max() should be after MinMaxScaler.<br> Then verify your predictions against the actual output.

```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
# here i have created my own dataframe
df = pd.DataFrame({"age":[20,23,25,19,29,27,45,32,78,64],
"income":[30000,40000,56000,18000,32000,26000,67000,120000,43000,55000],
"purchased":["yes","no","yes","no","yes","no","yes","yes","no","yes"]})
df
```

```python
df.columns # it's an attribute that gives us column names of a dataframe
df.values
```

```python
x = df[["age","income"]] # it is also way to slice dataframe , loc and iloc are other ways to do it.
```

```python
df.describe()
```

```python
scaler = StandardScaler() # standard scaler from scikit learn, uses population mean  but pandas std dev uses sample mean 
# scaler.fit(df[["age","income"]])
scaler.fit(df.iloc[:,0:2])
```

```python
df_scaled = scaler.transform(df[["age","income"]])
```

```python
scaler.mean_
```

```python
df_scaled = pd.DataFrame(df_scaled, columns=x.columns)
```

```python
df_scaled
```

```python
df_scaled['age'].describe()  # pandas .std() and the std shown by .describe() use: ddof = 1
```

```python
df_scaled['age'].std(ddof=0) # standard scaler uses population standard deviation
```

```python
df_scaled['age'].mean() # we may see and say, mathematically in z-score normalization, the mean of scaled column of 'age' should equal 0.
                        # but in the output we find, it is not exactly zero, but almost close to zero. that error is just representatin error of computers.
                        # computers store decimal numbers using floating-point binary representation. Some decimal calculations cannot be represented perfectly, so tiny rounding errors appear.
```

```python
round(df_scaled['age'].mean(),10) # the error is so small.
```
