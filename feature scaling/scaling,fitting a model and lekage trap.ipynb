```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("Social_Network_Ads.csv")
df = df[["Age","EstimatedSalary","Purchased"]]
ads = df[["Age","EstimatedSalary"]]
```

## Before Splitting

```python
scaler = StandardScaler()
print("Mean Before Scaling:")
print(ads.mean(),"\n")
print("Variance Before Scaling:")
print(ads.var())
# approach(a)
# fit the scaler on the entire dataset before splitting,
scaler.fit(ads)
ads_scaled = scaler.transform(ads)
ads_scaled = pd.DataFrame(ads_scaled,columns=["Age","EstimatedSalary"])
print("\n Mean After Scaling:")
print(ads_scaled.mean(),"\n")
print(" Variance After Scaling:")
ads_scaled.var()
```

## After Splitting

```python
# Splitting the dataset
X_train,X_test,y_train,y_test = train_test_split(ads,df['Purchased'],test_size=0.3,random_state= 1)
```

```python
# approach(b)
# fitting the scalar only with the training set
scaler.fit(X_train)
# and transforming both train and test set with the fitted parameters.
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```
