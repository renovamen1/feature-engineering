```python
class onehot_encoder():
    def __init__(self):
        pass
    def encode(self,column):
        self.column = column
        self.unique_values = list(column.value_counts().keys()) # getting the categories of a categorical column as a list
        
        ohe_cols = pd.DataFrame(
                    index=column.index,
                    columns= self.unique_values,
                    dtype = int 
                    ).fillna(0).astype(int)   #  before the inital values of the df was  in float, so to chnage that astype(int) is used here 

        for index,itr in self.column.items(): # this loop is for iterating over all the rows of the column to be encoded
            for unique_value in self.unique_values: 
                # if (itr == unique_value):
                ohe_cols.loc[index,itr] = 1   # this sets the value of matching colum to 1 .
                break
        return ohe_cols
```
